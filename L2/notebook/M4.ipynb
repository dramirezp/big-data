```python
import findspark
findspark.init()
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Basics').getOrCreate()
```

```python
import os
```

```python
files = os.listdir("../data/")[:10]
```

```python
print(files)
```

```python
df = spark.read.option("multiline", "true").json("../data/")
```

```python
df.count()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import explode
```

```python
df_rows = df.select(explode("collection"))
```

```python
df_rows.printSchema()
```

```python
df_rows = df.select(explode("collection").alias("rel"))
```

```python
df_rows.printSchema()
```

```python
df_rows.count()
```

```python
df_rows_schema = df_rows.select("rel.*")
```

```python
df_rows_schema.printSchema()
```

```python
df_rows_schema.show(10)
```

```python
df_rows_schema.select("rel_doi", "rel_title").show(10)
```

```python
df_rows_schema.select("rel_doi", "rel_title").show(10, False)
```

```python
df_author = df_rows_schema.select("rel_doi", "rel_authors")
```

```python
df_author.show(1, False)
```

```python
df_author.printSchema()
```

```python
df_authors_tmp = df_author.withColumn("author", explode("rel_authors"))
```

```python
df_authors_tmp.printSchema()
```

```python
from pyspark.sql.functions import col
```

```python
df_authors = df_authors_tmp.select(col("rel_doi").alias("doi"), col("author.author_name").alias("full_name"))
```

```python
df_authors.printSchema()
```

```python
df_authors.show(10, False)
```

```python
df_authors.count()
```

```python
df.createOrReplaceTempView("df")
```

```python
df_sql = spark.sql("""SELECT doi, author.author_name FROM (SELECT doi, explode(rel_authors) AS author FROM (SELECT item.rel_doi as doi, item.rel_authors FROM (SELECT explode(collection) as item from df)))""")
```

```python
df_sql.printSchema()
```

```python
df_sql.show(10, False)
```
